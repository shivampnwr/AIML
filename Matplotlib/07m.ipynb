```python
# Histogram:A visual representation of the distributionof the quantitative data.
#           they group values in bins(intervals) and count how many fall in each range
import matplotlib.pyplot as plt
import numpy as np
```

```python
scores=np.random.normal(loc=80,scale=10,size=100) # marks are centerd around 80,scale difference between marks
scores=np.clip(scores,0,100)# clip is used to bound the values like min or max values

plt.hist(scores,bins=10, color="brown",edgecolor="black")
plt.title("Exam Scores",fontsize=25,fontweight="bold")
plt.xlabel("Scores Obtained")
plt.ylabel("No. Of Students")
```
