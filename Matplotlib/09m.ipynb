```python
# matplotlib + Pandas
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df=pd.read_csv("pokemon.csv")
print(df)
```

```python
typecount=df["Type 1"].value_counts(ascending=True)
plt.barh(typecount.index,typecount.values,edgecolor="black",color="red")
plt.title("Pokemon types",fontweight="bold")
plt.xlabel("Count")
plt.ylabel("Type")
plt.tight_layout()
```
