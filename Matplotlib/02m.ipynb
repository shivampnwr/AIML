```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x=np.array([2023,2024,2025,2026])
y1=np.array([15,35,20,25])
y2=np.array([20,45,70,50])
```

```python
plt.plot(x,y1)
plt.plot(x,y2)
plt.title("Class Size",fontsize=25,
          family="Nimbus Sans",
          fontweight="bold",color="blue")
plt.xlabel("year",fontsize=20,
          family="Nimbus Sans",
          fontweight="bold",color="cyan")
plt.ylabel("gains",fontsize=20,
          family="Nimbus Sans",
          fontweight="bold",color="cyan")
plt.xticks(x)
plt.tick_params(axis="both",
               colors="magenta")
```
