```python
# Scattergraph: shows the relationship between two variables
#               helps identifying the corelation(+,-,none)
#               eg: study hours Vs test scores

import matplotlib.pyplot as plt
import numpy as np
```

```python
x1=[0,1,2,1,4,3,5,6,7,6]
y1=[34,45,55,40,75,64,81,87,93,91]

x2=[0,1,1,1,5,3,5,6,8,6]
y2=[34,45,55,40,75,44,81,87,90,89]

```

```python
plt.scatter(x1,y1,color="black",
            alpha=0.5,
            s=100,label="Class A")
plt.scatter(x2,y2,color="red",
            alpha=0.5,
            s=100,label="Class B")
plt.title("Scorecard")
plt.xlabel("Hours Studied")
plt.ylabel("Grades")
plt.legend()
```
