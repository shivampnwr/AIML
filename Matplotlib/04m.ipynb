```python
# bar charts: compare categories of data by representing each categories with a bar
import matplotlib.pyplot as plt
import numpy as np
```

```python
categories=np.array(["apple","orange","banana","grapes"])
values=np.array([1,7,15,0.5])
```

```python
# plt.bar(categories,values)
# '''for horizontal bargraph'''
plt.barh(categories,values)

plt.title("monthly fruit consumption",fontsize=25,fontstyle = "italic",fontweight="bold",color="orange")
plt.xlabel("fruits",fontsize=20,fontweight="bold",color="purple")
plt.ylabel("number in kg",fontsize=20,fontweight="bold",color="purple")
```
