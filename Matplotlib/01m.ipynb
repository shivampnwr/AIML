```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x=np.array([2023,2024,2025,2026])
y1=np.array([15,35,20,25])
y2=np.array([20,45,70,50])
```

```python
plt.plot(x,y1,marker='.',
         markersize=10,
         markerfacecolor="green",
         markeredgecolor="black",
         linestyle="dashdot",
         linewidth=2,
         color="#2b8554")
plt.plot(x,y2,marker='.',
         markersize=10,
         markerfacecolor="red",
         markeredgecolor="none",
         linestyle="dashdot",
         linewidth=2,
         color="#000000")
# we use mfc for markerfacecolor and we can use hexadecimal values for the color 
```

```python
# we can use dictionary to use the same customisation for different lines
line_styles=dict(marker='.',
         markersize=10,
         markerfacecolor="red",
         markeredgecolor="none",
         linestyle="dashdot",
         linewidth=2,
         color="#000000")
```

```python
plt.plot(x,y1,**line_styles)
plt.plot(x,y2,**line_styles)
```
